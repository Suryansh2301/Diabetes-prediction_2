# Diabetes Prediction Project
## EDA and Data Cleaning

### Objective

The objective of this project is to analyze a diabetes prediction dataset,
understand the relationships between patient characteristics and diabetes,
and prepare a clean dataset for machine learning model development.

### Dataset Features

- Gender
- Age
- Hypertension
- Heart Disease
- Smoking History
- BMI
- HbA1c Level
- Blood Glucose Level
- Diabetes (Target)
