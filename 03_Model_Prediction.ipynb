```python
import streamlit as st
import pandas as pd
import joblib
```

```python
# PAGE CONFIGURATION
st.set_page_config(
    page_title="Diabetes Detection",
    page_icon="🩺",
    layout="centered"
)
```

```python
# LOAD MODEL
@st.cache_resource
def load_model():

    model = joblib.load("diabetes_model.pkl")

    return model


model = load_model()
```

```python
# TITLE
st.title("🩺 Diabetes Detection System")

st.write(
    """
    Enter the patient's information below to estimate the
    probability of diabetes.
    """
)

st.divider()
```

```python
# USER INPUT
st.subheader("Patient Information")
```

```python
# Gender
gender = st.selectbox(
    "Gender",
    options=[
        "Female",
        "Male",
        "Other"
    ]
)
```

```python
# Age
age = st.number_input(
    "Age",
    min_value=0.0,
    max_value=120.0,
    value=30.0,
    step=1.0
)
```

```python
# Hypertension
hypertension = st.selectbox(
    "Hypertension",
    options=[0, 1],
    format_func=lambda x:
        "No" if x == 0 else "Yes"
)
```

```python
# Heart Disease
heart_disease = st.selectbox(
    "Heart Disease",
    options=[0, 1],
    format_func=lambda x:
        "No" if x == 0 else "Yes"
)
```

```python
# Smoking History
smoking_history = st.selectbox(
    "Smoking History",
    options=[
        "never",
        "former",
        "current",
        "not current",
        "ever",
        "No Info"
    ]
)
```

```python
# BMI
bmi = st.number_input(
    "BMI",
    min_value=1.0,
    max_value=100.0,
    value=25.0,
    step=0.1
)
```

```python
# HbA1c
hba1c = st.number_input(
    "HbA1c Level",
    min_value=1.0,
    max_value=20.0,
    value=5.5,
    step=0.1
)
```

```python
# Blood Glucose
blood_glucose = st.number_input(
    "Blood Glucose Level",
    min_value=1,
    max_value=500,
    value=100,
    step=1
)


st.divider()
```

```python
# PREDICTION
if st.button(
    "🔍 Predict Diabetes",
    use_container_width=True
):

    # Create input DataFrame
    input_data = pd.DataFrame({
        "gender": [gender],
        "age": [age],
        "hypertension": [hypertension],
        "heart_disease": [heart_disease],
        "smoking_history": [smoking_history],
        "bmi": [bmi],
        "HbA1c_level": [hba1c],
        "blood_glucose_level": [blood_glucose]
    })

    # Prediction
    prediction = model.predict(input_data)[0]

    # Probability
    probability = model.predict_proba(input_data)[0][1]

    st.divider()
    st.subheader("Prediction Result")

    if prediction == 1:
        st.error("Prediction: Diabetes")
    else:
        st.success("Prediction: No Diabetes")

    st.metric(
        "Diabetes Probability",
        f"{probability * 100:.2f}%"
    )

    st.subheader("Patient Information")

    st.dataframe(
        input_data,
        use_container_width=True
    )
```
